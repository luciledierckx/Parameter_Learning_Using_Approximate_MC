```python
import pandas as pd
import os
import numpy as np


ds_types = ['bayesian_networks']
table = pd.DataFrame(columns=['dataset', "$\epsilon=0.0$", "$\epsilon=0.1$", "$\epsilon=0.5$", "$\epsilon=0.8$"])
#table['dataset'] = ["sampling", "$\epsilon = 0.0$", "$\epsilon = 0.05$", "$\epsilon = 0.3$", "$\epsilon = 0.8$", "$\epsilon = 0.05, \delta = 0.2$", "$\epsilon = 0.05, \delta = 0.5$", "$\epsilon = 0.05, \delta = 0.8$", "$\epsilon = 0.3, \delta = 0.2$", "$\epsilon = 0.3, \delta = 0.5$", "$\epsilon = 0.3, \delta = 0.8$", "$\epsilon = 0.8, \delta = 0.2$", "$\epsilon = 0.8, \delta = 0.5$", "$\epsilon = 0.8, \delta = 0.8$"]

for ds_type in ds_types:
    path = "/home/ldierckx/outputs_NeSy/"
    path += ds_type
    if ds_type== "bayesian_networks":
        datasets = ["munin"]#['alarm', 'andes', 'asia', 'cancer', 'child', 'earthquake', 'hailfinder', 'hepar2', 'insurance', 'munin', 'pathfinder', 'pigs', 'sachs', 'survey', 'water', 'win95pts']
    
    for dataset in datasets:
        line = [dataset]
        try:                
            # Epsilon
            for subfolder in ['epsilon0.0', 'epsilon0.1', 'epsilon0.5', 'epsilon0.8']:
                folder = path+"/"+dataset+"/train_test_size/"+subfolder+"/"
                df = pd.read_csv(folder+sorted(os.listdir(folder))[-1])
                first = 0
                last = 0
                for iter in range(0, 10):
                    df_iter = df[df['iter'] == iter]
                    first += (df_iter['avg_errror'].iloc[0])
                    last += (df_iter['avg_errror'].iloc[-1])
                line.append('%.1e $\\rightarrow$ %.1e' % (first/10, last/10))

            table.loc[len(table)] = line
        except Exception as e:
            print(e)
            continue
print(table.to_latex(index=False))
```
