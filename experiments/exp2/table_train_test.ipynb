```python
import pandas as pd
import os
import numpy as np


ds_types = ['bayesian_networks']
table = pd.DataFrame()
table[''] = ["$\\varepsilon = 0.0$", "$\\varepsilon = 0.05$", "$\\varepsilon = 0.3$", "$\\varepsilon = 0.8$", "$\\varepsilon = 0.05, \delta = 0.2$", "$\\varepsilon = 0.05, \delta = 0.5$", "$\\varepsilon = 0.05, \delta = 0.8$", "$\\varepsilon = 0.3, \delta = 0.2$", "$\\varepsilon = 0.3, \delta = 0.5$", "$\\varepsilon = 0.3, \delta = 0.8$", "$\\varepsilon = 0.8, \delta = 0.2$", "$\\varepsilon = 0.8, \delta = 0.5$", "$\\varepsilon = 0.8, \delta = 0.8$"]

for ds_type in ds_types:
    path = "/experiments/outputs_NeSy/"
    path += ds_type
    if ds_type== "bayesian_networks":
        datasets = ['munin', 'hepar2', "water"]#['alarm', 'andes', 'asia', 'cancer', 'child', 'earthquake', 'hailfinder', 'hepar2', 'insurance', 'munin', 'pathfinder', 'pigs', 'sachs', 'survey', 'water', 'win95pts']
    
    for dataset in datasets:
        line = []
        try:            
            # Epsilon
            for subfolder in ['epsilon0.0', 'epsilon0.05', 'epsilon0.3', 'epsilon0.8']:
                folder = path+"/"+dataset+"/train_test/"+subfolder+"/"
                df = pd.read_csv(folder+sorted(os.listdir(folder))[-1])
                first = 0
                last = 0
                for iter in range(0, 10):
                    df_iter = df[df['iter'] == iter]
                    first += (df_iter['avg_errror'].iloc[0])
                    last += (df_iter['avg_errror'].iloc[-1])
                line.append('%.1e $\\rightarrow$ %.1e' % (first/10, last/10))
                    
                
            # Epsilon-Delta
            for  e in [0.05, 0.3, 0.8]:
                for d in [0.2, 0.5, 0.8]:
                    folder = path+"/"+dataset+"/train_test/epsilon"+str(e)+"delta"+str(d)+"/"
                    df = pd.read_csv(folder+sorted(os.listdir(folder))[-1])
                    first = 0
                    last = 0
                    for iter in range(0, 10):
                        df_iter = df[df['iter'] == iter]
                        first += (df_iter['avg_errror'].iloc[0])
                        last += (df_iter['avg_errror'].iloc[-1])
                    line.append('%.1e $\\rightarrow$ %.1e' % (first/10, last/10))
            table[dataset] = line
        except Exception as e:
            print(e)
            continue
print(table.to_latex(index=False))
```
