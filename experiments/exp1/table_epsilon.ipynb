```python
import pandas as pd
import os
import numpy as np

ds_types = ["bayesian_networks", "power_grids", "water_networks"]
table = pd.DataFrame()#(columns=["dataset", "sampling", "$\epsilon = 0.0$", "$\epsilon = 0.05$", "$\epsilon = 0.3$", "$\epsilon = 0.8$", "$\epsilon = 0.05, \delta = 0.2$", "$\epsilon = 0.05, \delta = 0.5$", "$\epsilon = 0.05, \delta = 0.8$", "$\epsilon = 0.3, \delta = 0.2$", "$\epsilon = 0.3, \delta = 0.5$", "$\epsilon = 0.3, \delta = 0.8$", "$\epsilon = 0.8, \delta = 0.2$", "$\epsilon = 0.8, \delta = 0.5$", "$\epsilon = 0.8, \delta = 0.8$"])
table['dataset'] = ["sampling", "$\epsilon = 0.0$", "$\epsilon = 0.05$", "$\epsilon = 0.3$", "$\epsilon = 0.8$", "$\epsilon = 0.05, \delta = 0.2$", "$\epsilon = 0.05, \delta = 0.5$", "$\epsilon = 0.05, \delta = 0.8$", "$\epsilon = 0.3, \delta = 0.2$", "$\epsilon = 0.3, \delta = 0.5$", "$\epsilon = 0.3, \delta = 0.8$", "$\epsilon = 0.8, \delta = 0.2$", "$\epsilon = 0.8, \delta = 0.5$", "$\epsilon = 0.8, \delta = 0.8$"]
    
for ds_type in ds_types:
    path = "/experiments/outputs_NeSy/"
    path += ds_type+"/"
    if ds_type== "bayesian_networks":
        datasets = ['munin']#['alarm', 'andes', 'asia', 'cancer', 'child', 'earthquake', 'hailfinder', 'hepar2', 'insurance', 'munin', 'pathfinder', 'pigs', 'sachs', 'survey', 'water', 'win95pts']
    elif ds_type == "power_grids":
        datasets = ['Tennessee']#['Alabama', 'Alaska', 'Albania', 'Arizona', 'Arkansas', 'Armenia', 'Austria', 'Belarus', 'Belgium', 'Bosnia_and_Herzegovina', 'Bulgaria', 'California', 'Connecticut', 'Croatia', 'Czech_Republic', 'Delaware', 'Denmark', 'Estonia', 'Finland', 'Florida', 'France', 'Georgia_eu', 'Georgia_us', 'Germany', 'Greece', 'Hungary', 'Iceland', 'Idaho', 'Illinois', 'Iowa', 'Ireland', 'Italy', 'Kansas', 'Kentucky', 'Latvia', 'Lithuania', 'Louisiana', 'Luxembourg', 'Maine', 'Maryland', 'Massachusetts', 'Michigan', 'Minnesota', 'Mississippi', 'Missouri', 'Montana', 'Montenegro', 'Nebraska', 'Netherlands', 'Nevada', 'New_Hampshire', 'New_Jersey', 'New_Mexico', 'New_York', 'North_Carolina', 'North_Dakota', 'Norway', 'Ohio', 'Oklahoma', 'Oregon', 'Pennsylvania', 'Poland', 'Portugal', 'Republic_of_Moldova', 'Rhode_Island', 'Romania', 'Russia', 'Serbia', 'Slovakia', 'Slovenia', 'South_Carolina', 'South_Dakota', 'Spain', 'Sweden', 'Switzerland', 'Tennessee', 'Texas', 'The_former_Yugoslav_Republic_of_Macedonia', 'Turkey', 'Ukraine', 'United_Kingdom', 'Utah', 'Vermont', 'Virginia', 'Washington', 'West_Virginia', 'Wisconsin', 'Wyoming']
    elif ds_type == "water_networks":
        datasets = ["Net3"]#['Net1', 'Net2', 'Net3', 'Net6']
    for dataset in datasets:
        line = []
        try:
            # Sampling
            sampl_folder = path+dataset+"/approx_vary/sampling/"
            results = sampl_folder + sorted(os.listdir(sampl_folder))[-1]
            df = pd.read_csv(results, header=0, sep=',')
            final_loss = []
            for repeat in range(10):
                final_loss.append(df[df['iter']==repeat]['avg_errror'].iloc[-1])
            line.append('%.1e' % np.mean(final_loss) + " $\pm$ " + '%.1e' % np.std(final_loss))
            
            # Epsilon
            for e in [0.0, 0.05, 0.3, 0.8]:
                folder = path+dataset+"/approx_vary/epsilon"+str(e)+"/"
                results = folder + sorted(os.listdir(folder))[-1]
                df = pd.read_csv(results, header=0, sep=',')
                final_loss = []
                for repeat in range(10):
                    final_loss.append(df[df['iter']==repeat]['avg_errror'].iloc[-1])
                line.append('%.1e' % np.mean(final_loss) + " $\pm$ " + '%.1e' % np.std(final_loss))
                
            # Epsilon Delta
            for e in [0.05, 0.3, 0.8]:
                for d in [0.2, 0.5, 0.8]:
                    folder = path+dataset+"/approx_vary/epsilon"+str(e)+"delta"+str(d)+"/"
                    results = folder + sorted(os.listdir(folder))[-1]
                    df = pd.read_csv(results, header=0, sep=',')
                    final_loss = []
                    for repeat in range(10):
                        final_loss.append(df[df['iter']==repeat]['avg_errror'].iloc[-1])
                    line.append('%.1e' % np.mean(final_loss) + " $\pm$ " + '%.1e' % np.std(final_loss))
            #table.loc[len(table)] = line
            table[dataset] = line

        except Exception as e:
            print(e)
            continue
# Print the table in a LaTeX format
print(table.to_latex(index=False))#, columns=["dataset", "sampling", "$\epsilon = 0.0$", "$\epsilon = 0.05$", "$\epsilon = 0.3$", "$\epsilon = 0.8$", "$\epsilon = 0.05, \delta = 0.2$", "$\epsilon = 0.05, \delta = 0.5$", "$\epsilon = 0.05, \delta = 0.8$", "$\epsilon = 0.3, \delta = 0.2$", "$\epsilon = 0.3, \delta = 0.5$", "$\epsilon = 0.3, \delta = 0.8$", "$\epsilon = 0.8, \delta = 0.2$", "$\epsilon = 0.8, \delta = 0.5$", "$\epsilon = 0.8, \delta = 0.8$"]))
```
